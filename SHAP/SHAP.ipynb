```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel(
#    "XXX_SHAP_values_all_targets.xlsx",   # XXX is name of the gregression
    sheet_name="Mean_abs_SHAP_summary"
)

targets = df["Target"].unique()

feature_colors = {
    "MA": "#285A48",
    "FA": "#408A71",
    "Cs": "#B0E4CC",
    "EA": "#AEB784",
    "Pb": "#E68457",
    "Sn": "#FFE8B4",
    "Cl": "#F1E9E9",
    "Br": "#E491C9",
    "I":  "#982598"
}
```

```python
for target in targets:

    temp = df[df["Target"] == target]

    colors = [feature_colors[f] for f in temp["Feature"]]

    plt.figure(figsize=(6, 6))

    plt.bar(
        temp["Feature"],
        temp["Mean_abs_SHAP"],
        color=colors,
        alpha=0.8,
        edgecolor="black",
        linewidth=1.5
    )

    plt.ylabel("Mean |SHAP|", fontsize=18)
    plt.xlabel("Feature", fontsize=18)
    plt.title(target, fontsize=18)

    if target == "Efficiency (%)":
        plt.ylim(0, 2.0)
        plt.yticks([0, 0.5, 1.0, 1.5, 2.0])

    elif target == "Open circuit voltage (V)":
        plt.ylim(0, 0.10)
        plt.yticks([0, 0.02, 0.04, 0.06, 0.08, 0.10])

    elif target == "Short circuit current density (A/m²)":
        plt.ylim(0, 25)
        plt.yticks([0, 5, 10, 15, 20, 25])

    elif target == "Fill factor":
        plt.ylim(0, 0.030)
        plt.yticks([0, 0.005, 0.010, 0.015, 0.020, 0.025, 0.030])

    elif target == "Band gap (eV)":
        plt.ylim(0, 0.15)
        plt.yticks([0, 0.03, 0.06, 0.09, 0.12, 0.15])

    plt.xticks(fontsize=18)
    plt.yticks(fontsize=18)
    
    plt.tick_params(
    axis="both",
    width=2,
    length=6
    )
    
    ax = plt.gca()
    ax.spines["left"].set_linewidth(2)
    ax.spines["right"].set_linewidth(2)
    ax.spines["top"].set_linewidth(2)
    ax.spines["bottom"].set_linewidth(2)

    plt.tight_layout()

    filename = target.replace(" ", "_")
    filename = filename.replace("(", "").replace(")", "")
    filename = filename.replace("/", "_")
    filename = filename.replace("²", "2")

    plt.savefig(
    f"{filename}_Mean_abs_SHAP.png",
    dpi=600,
    bbox_inches="tight"
    )
```
